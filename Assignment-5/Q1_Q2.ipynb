```python
#Q1

from PIL import Image
import numpy as np
from google.colab import files
import io
import matplotlib.pyplot as plt

# Upload image
uploaded = files.upload()

# Read grayscale image
filename = next(iter(uploaded))
img = Image.open(io.BytesIO(uploaded[filename])).convert("L")

# Convert image to NumPy array
image = np.array(img)

# Calculate histogram
hist = np.bincount(image.flatten(), minlength=256)

# Calculate cumulative distribution function
cdf = hist.cumsum()

# Normalize CDF
cdf_min = cdf[cdf > 0].min()
cdf_normalized = (cdf - cdf_min) / (image.size - cdf_min)

# Create lookup table
lookup_table = np.clip(cdf_normalized * 255, 0, 255).astype(np.uint8)

# Apply histogram equalization
equalized_image = lookup_table[image]

# Convert to PIL image
result = Image.fromarray(equalized_image)

# Display original and equalized images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized_image, cmap="gray")
plt.title("Equalized Image")
plt.axis("off")

plt.show()

# Save result
result.save("histogram_equalized.jpg")

print("Histogram Equalization Completed!")

```

```python
#Q2
from PIL import Image
import numpy as np
from google.colab import files
import io
import matplotlib.pyplot as plt

# Upload source image
print("Upload Source Image:")
source_uploaded = files.upload()

source_name = next(iter(source_uploaded))
source_img = Image.open(
    io.BytesIO(source_uploaded[source_name])
).convert("L")

# Upload reference image
print("Upload Reference Image:")
reference_uploaded = files.upload()

reference_name = next(iter(reference_uploaded))
reference_img = Image.open(
    io.BytesIO(reference_uploaded[reference_name])
).convert("L")

# Convert images to NumPy arrays
source = np.array(source_img)
reference = np.array(reference_img)

# Calculate source histogram
source_hist = np.bincount(
    source.flatten(), minlength=256
)

# Calculate reference histogram
reference_hist = np.bincount(
    reference.flatten(), minlength=256
)

# Calculate cumulative distribution functions
source_cdf = source_hist.cumsum()
reference_cdf = reference_hist.cumsum()

# Normalize CDF
source_cdf = source_cdf / source.size
reference_cdf = reference_cdf / reference.size

# Create mapping table
lookup_table = np.zeros(256, dtype=np.uint8)

for i in range(256):
    difference = np.abs(reference_cdf - source_cdf[i])
    lookup_table[i] = np.argmin(difference)

# Apply histogram matching
matched_image = lookup_table[source]

# Convert to PIL image
result = Image.fromarray(matched_image)

# Display images
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(source, cmap="gray")
plt.title("Source Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(reference, cmap="gray")
plt.title("Reference Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(matched_image, cmap="gray")
plt.title("Histogram Matched Image")
plt.axis("off")

plt.show()

# Save result
result.save("histogram_matched.jpg")

print("Histogram Matching Completed!")
```
