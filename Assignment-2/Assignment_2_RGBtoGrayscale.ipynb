```python
import cv2
import numpy as np

# Read RGB (color) image
img = cv2.imread("image.jpg")

if img is None:
    print("Error: Image not found!")
    exit()

# OpenCV reads image as BGR, so convert it to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Separate R, G and B planes
R = rgb[:, :, 0].astype(float)
G = rgb[:, :, 1].astype(float)
B = rgb[:, :, 2].astype(float)


# ---------------- CASE 1 ----------------
# Mean average of three planes

gray_mean = ((R + G + B) / 3).astype(np.uint8)

cv2.imshow("Original RGB Image", img)
cv2.imshow("Grayscale - Mean Average", gray_mean)

cv2.waitKey(0)
cv2.destroyAllWindows()


# ---------------- CASE 2 ----------------
# Input weightage of R, G and B from user

print("\nEnter weightages for R, G and B")
print("Weightages should be between 0 and 1")
print("Sum of all weightages should be equal to 1")

wr = float(input("Enter weightage of R: "))
wg = float(input("Enter weightage of G: "))
wb = float(input("Enter weightage of B: "))

# Check weightages
if not (0 <= wr <= 1 and 0 <= wg <= 1 and 0 <= wb <= 1):
    print("Error: Weightages must be between 0 and 1.")
    exit()

if not np.isclose(wr + wg + wb, 1.0):
    print("Error: Sum of weightages must be equal to 1.")
    exit()

# Weighted grayscale conversion
gray_weighted = (
    wr * R +
    wg * G +
    wb * B
).astype(np.uint8)

cv2.imshow("Grayscale - Weighted Average", gray_weighted)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
